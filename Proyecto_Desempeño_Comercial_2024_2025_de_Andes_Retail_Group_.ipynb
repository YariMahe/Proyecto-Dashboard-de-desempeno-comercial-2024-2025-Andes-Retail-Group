### Narrativa del Dashboard (Modelo SCQA)

#### 🖥️ Vista General (Overview)

**S (Situación):**  
La empresa Andes Retail mantiene una operación estable con ingresos similares en los años 2024 (2,863,384) y 2025 (2,668,610).

**C (Complicación):**  
A pesar de eso en 2025 hubieron menos ganancias (935,937.71) respecto a 2024 (1,005,811.62), representando una casi del 7% aproximadamente.

**Q (Pregunta):**   
¿Se debe a menos unidades vendidas?

**A (Respuesta):**  
Comparando las unidades vendidas en 2024 con las del 2025, las primeras fueron mayores: 29,496 respecto a 28,175 de 2025. Aunque los costos también bajaron, la caída en ingresos fue mayor (194,774) que la caída en los costos (124,900), es por ello que la ganancia fue menor.

---

#### 🔎 Vista Detalle

**S (Situación):**  
Los ingresos de la empresa Andes Retail son más altos en la estación "Verano" con 2.24 millones, seguido de "Otoño" con 1.34 millones, "Primavera" con 1.29 millones y por último "Invierno" con 0.65 millones. Respecto a la segmentación de clientes, "Premium" domina con 2.60 millones, después le sigue "Estándar" con 2.44 millones, dejando a "Económico" en último lugar con 0.50 millones.



**C (Complicación):**  
Sin embargo, el total de ingresos respecto a las unidades vendidas demuestra que las unidades vendidas por sí solas no determinan ingresos más altos, incluso los ingresos tienden a descender a partir de 18 unidades adquiridas aproximadamente.



**Q (Pregunta):**  
¿Qué implica esto para la estrategia de ventas?

**A (Respuesta):**  
Las estrategias de marketing deberían enfocarse en clientes que compran como máximo 17 productos sobre todo de los sectores "Premium" y "Estándar", así como aprovechar la temporada de "Verano" que es donde se generan los mayores ingresos.
